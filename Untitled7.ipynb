```python
import pandas as pd

from google.colab import files
uploaded = files.upload()

df = pd.read_csv('play_tennis.csv')
print(df)
```

```python
print("First 5 rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nDataset shape:")
print(df.shape)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Remove day column
data = df.drop('day', axis=1)

# Convert text values into numbers
encoders = {}

for col in data.columns:
    encoders[col] = LabelEncoder()
    data[col] = encoders[col].fit_transform(data[col])

# Separate input and output
X = data.drop('play', axis=1)
y = data['play']

# Train the model
model = DecisionTreeClassifier(criterion='entropy')
model.fit(X, y)

print("Model trained successfully!")
```

```python
plt.figure(figsize=(12, 8))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=encoders['play'].classes_,
    filled=True
)

plt.show()
```

```python
# Example: Outlook=Sunny, Temp=Cool,
# Humidity=High, Wind=Strong

sample = pd.DataFrame(
    [['Sunny', 'Cool', 'High', 'Strong']],
    columns=['outlook', 'temp', 'humidity', 'wind']
)

# Convert values using the trained encoders
for col in sample.columns:
    sample[col] = encoders[col].transform(sample[col])

prediction = model.predict(sample)

print("Play Tennis Prediction:", encoders['play'].inverse_transform(prediction)[0])
```
