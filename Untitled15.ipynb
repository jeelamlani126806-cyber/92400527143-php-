```python
import pandas as pd
```

```python
data = {
    'Age' : ['Youth','Youth','Middle','Senior','senior','Middle','Senior','Youth','Youth','Senior','Youth'],
    'Income' : ['High','High','High','Medium','Low','Low','low','Medium','Low','Medium','Medium'],
    'Student' : ['No','No','No','No','Yes','Yes','Yes','No','Yes','Yes','Yes'],
    'Credit_rating' : ['Fair','Excellent','Fair','Fair','Fair','Excellent','Excellent','Fair','Fair','Fair','Excellent'],
    'Buys_computer' : ['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes','Yes']

}
```

```python
df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoders = {}
for column in df.columns:
    le = LabelEncoder()
    df[column] = le.fit_transform(df[column])
    label_encoders[column] = le
```

```python
X = df.drop('Buys_computer', axis=1)
y = df['Buys_computer']
```

```python
X
```

```python
y
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X, y)
```

```python
print('Now model is trained')
```

```python
new_data = pd.DataFrame({
    'Age' : ['youth'],
    'Income' : ['Merdium'],
    'Student' : ['yes'],
    'Credit_rating' : ['Fair']
})
```

```python
for col in new_data.columns :
  new_data[col] = label_encoders[col].transform(new_data[col])
```
