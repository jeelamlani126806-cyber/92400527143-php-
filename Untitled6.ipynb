```python
import pandas as pd
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("/content/pima-indians-diabetes.csv")
df
```

```python

le = {} #CREATE EMPTY DICTIONARY
df_enc = df.copy() # here we just cop our original data
for col in df.columns: # it convert into the text using loop through of the column
    le[col] = LabelEncoder()  # now create new label for curent data
    df_enc[col] = le[col].fit_transform(df[col])

```

```python
# Separate features and target class from the encoded DataFrame
# The last column '1' is the target column in your loaded dataset
x = df_enc.drop(columns=['1'])
y = df_enc['1']

model = CategoricalNB() # apply model
model.fit(x, y) # here fit your values
```

```python
# Define a raw test sample
test_raw = [1, 85, 66, 29, 0, 26.6, 0.351, 31]

# Get the feature columns (all columns except the target class '1')
feature_cols = [col for col in df.columns if col != '1']

# Convert each raw value in the test sample using its respective LabelEncoder
test_encoded = []
for val, col in zip(test_raw, feature_cols):
    # We use transform to encode the test data based on what was learned during training
    encoded_val = le[col].transform([val])[0]
    test_encoded.append(encoded_val)

print("\n----prediction-------------------------")
# Predict using the encoded test sample
pred = model.predict([test_encoded])[0]
prob = model.predict_proba([test_encoded])[0]

# Inverse transform to get the original target value
orig_pred = le['1'].inverse_transform([pred])[0]

print(f"Raw Input: {test_raw}")
print(f"Encoded Input: {test_encoded}")
print(f"Result (Diabetes Outcome): {orig_pred}")
print(f"Probability of Class 0: {prob[0]:.4f}")
print(f"Probability of Class 1: {prob[1]:.4f}")
```

```python
from sklearn.metrics import accuracy_score
y_pred = model.predict(x)
print(f"\nTraining Accuracy:{accuracy_score(y, y_pred)*100:.2f}%")

```
