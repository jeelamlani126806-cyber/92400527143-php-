```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
data = {
    "Weather" : ["Sunny","Cloudy","Sunny","Cloudy","Rainy","Rainy","Rainy","Sunny","Cloudy","Rainy"],
    "Temperature" : ["Hot","Hot","Mild","Mild","Mild","Cool","Mild","Hot","Hot","Mild"],
    "Humidity" : ["High","High","Normal","High","High","Normal","High","High","Normal","High"],
    "Wind" : ["Weak","Weak","Strong","Strong","Strong","Strong","Weak","Strong","Weak","Strong"],
    "Play" : ["No","Yes","Yes","Yes","No","No","Yes","No","Yes","No"]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df_encoded =  df.replace({"Sunny": 0, "Cloudy" : 1, "Rainy":2,
                          "Hot" : 0, "Mild" : 1, "Cool" : 2,
                          "High" :0, " Normal" : 1,
                          "Weak" : 0, "Strong" : 1,
                          "No" : 0, "Yes" : 1})
```

```python
# Fix the "Normal" encoding typo if it exists
df_encoded["Humidity"] = df_encoded["Humidity"].replace("Normal", 1)

# Correcting the case-sensitive column name "Play"
X = df_encoded.drop("Play", axis=1)
Y = df_encoded["Play"]
```

```python
model = DecisionTreeClassifier()
model.fit(X,Y)
```

```python
print ("\n Model Trained")
```

```python
sample = [[1,1,0,0]]
prediction = model . predict(sample)
```
